```python
import cv2
import matplotlib.pyplot as plt

paths = [
    "/kaggle/input/datasets/mithilamostofa/cse438-pic/Image1.jpeg",
    "/kaggle/input/datasets/mithilamostofa/cse438-pic/Image2.jpeg",
    "/kaggle/input/datasets/mithilamostofa/cse438-pic/Image3.jpeg",
    "/kaggle/input/datasets/mithilamostofa/cse438-pic/Image4.jpeg"
]

images = []

for path in paths:
    img = cv2.imread(path)
    img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
    images.append(img)

# Show images
for i, img in enumerate(images):
    plt.imshow(img)
    plt.title(f"Image {i+1}")
    plt.axis('off')
    plt.show()
```

```python
import numpy as np

img = images[3]  # first image

# reshape image into list of pixels
pixels = img.reshape(-1, 3)

# sort by RED channel (index 0), descending
sorted_pixels = pixels[pixels[:, 0].argsort()[::-1]]

# reshape back to image
sorted_img = sorted_pixels.reshape(img.shape)

# show result
plt.imshow(sorted_img)
plt.title("Red Channel Sorted")
plt.axis('off')
plt.show()
```

```python
img = images[0].copy()

# rotate
rotated = np.rot90(img)

# column sort
for col in range(rotated.shape[1]):
    col_pixels = rotated[:, col, :]
    intensity = col_pixels.mean(axis=1)
    sorted_col = col_pixels[intensity.argsort()[::-1]]
    rotated[:, col, :] = sorted_col

# rotate back
final_img = np.rot90(rotated, k=3)

plt.imshow(final_img)
plt.title("45 Degree Sorted")
plt.axis('off')
```

```python
import numpy as np
import matplotlib.pyplot as plt

img = images[0].copy()
h, w, c = img.shape

# padding
size = max(h, w)
padded = np.zeros((size, size, 3), dtype=img.dtype)
padded[:h, :w] = img

# diagonal sorting
for k in range(-size+1, size):
    diag_pixels = []

    for i in range(size):
        j = i - k
        if 0 <= j < size:
            diag_pixels.append(padded[i, j])

    diag_pixels = np.array(diag_pixels)

    # intensity sort
    intensity = diag_pixels.mean(axis=1)
    sorted_diag = diag_pixels[intensity.argsort()[::-1]]

    idx = 0
    for i in range(size):
        j = i - k
        if 0 <= j < size:
            padded[i, j] = sorted_diag[idx]
            idx += 1

# crop back
final_img = padded[:h, :w]

# 🔥 SHOW SIDE BY SIDE
plt.figure(figsize=(10,5))

plt.subplot(1,2,1)
plt.imshow(img)
plt.title("Original Image")
plt.axis('off')

plt.subplot(1,2,2)
plt.imshow(final_img)
plt.title("45° Diagonal Sorted (Padding)")
plt.axis('off')

plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

img = images[0].copy()
h, w, _ = img.shape

center_x, center_y = h // 2, w // 2

pixel_list = []

# collect pixel info (per pixel)
for i in range(h):
    for j in range(w):
        pixel = img[i, j]
        
        # distance from center
        dist = (i - center_x)**2 + (j - center_y)**2
        
        # intensity
        intensity = pixel.mean()
        
        # store everything
        pixel_list.append((dist, intensity, pixel))

# sort: first by distance, then by intensity
pixel_list.sort(key=lambda x: (x[0], -x[1]))

# rebuild image
new_img = np.zeros_like(img)
idx = 0

for i in range(h):
    for j in range(w):
        new_img[i, j] = pixel_list[idx][2]
        idx += 1

# show side by side
plt.figure(figsize=(10,5))

plt.subplot(1,2,1)
plt.imshow(img)
plt.title("Original")
plt.axis('off')

plt.subplot(1,2,2)
plt.imshow(new_img)
plt.title("Center-wise Sorted (Per Pixel)")
plt.axis('off')

plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

img = images[0]
h, w, _ = img.shape

# flatten pixels
pixels = img.reshape(-1, 3)

# sort pixels by intensity (bright → dark)
intensity = pixels.mean(axis=1)
sorted_pixels = pixels[intensity.argsort()[::-1]]

# create empty image
new_img = np.zeros_like(img)

# center
cx, cy = h // 2, w // 2

# create list of coordinates with distance
coords = []
for i in range(h):
    for j in range(w):
        dist = (i - cx)**2 + (j - cy)**2
        coords.append((dist, i, j))

# sort coordinates (center → outward)
coords.sort(key=lambda x: x[0])

# fill pixels
for idx, (_, i, j) in enumerate(coords):
    new_img[i, j] = sorted_pixels[idx]

# show side by side
plt.figure(figsize=(10,5))

plt.subplot(1,2,1)
plt.imshow(img)
plt.title("Original Image")
plt.axis('off')

plt.subplot(1,2,2)
plt.imshow(new_img)
plt.title("Center-wise Sorted (Radial)")
plt.axis('off')

plt.show()
```
